```python
import matplotlib.pyplot as plt
from collections import Counter

name = "SAHIL"

# Count letters
letters = Counter(name)

# Data for graph
x = list(letters.keys())
y = list(letters.values())

# Plot graph
plt.bar(x, y)
plt.title("Alphabet Frequency in Name 'SAHIL'")
plt.xlabel("Alphabets")
plt.ylabel("Count")
plt.show()
```
